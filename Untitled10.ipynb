```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
from IPython.display import display

# Upload all 5 CSV files together
uploaded = files.upload()

reviews = pd.read_csv("reviews.csv")
products = pd.read_csv("products.csv")
sellers = pd.read_csv("sellers.csv")
orders = pd.read_csv("orders.csv")
support = pd.read_csv("support.csv")

print("All five datasets loaded successfully!")
```

```python

datasets = {
    "Reviews": reviews,
    "Products": products,
    "Sellers": sellers,
    "Orders": orders,
    "Support": support
}

for name, df in datasets.items():
    print("\n==========", name, "==========")
    print("Shape:", df.shape)
    print("\nFirst 5 rows:")
    display(df.head())
    print("\nColumn names:")
    print(df.columns.tolist())
    print("\nData types:")
    print(df.dtypes)
    print("\nMissing values:")
    print(df.isnull().sum())
    print("\nDuplicate rows:", df.duplicated().sum())
```

```python

# Make copies so the original data remains unchanged
reviews_clean = reviews.copy()
products_clean = products.copy()
sellers_clean = sellers.copy()
orders_clean = orders.copy()
support_clean = support.copy()

# Remove duplicate reviews using review_id if available
if "review_id" in reviews_clean.columns:
    reviews_clean = reviews_clean.drop_duplicates(
        subset=["review_id"], keep="first"
    )
else:
    reviews_clean = reviews_clean.drop_duplicates()

# Standardize product category labels
if "category" in products_clean.columns:
    products_clean["category"] = (
        products_clean["category"]
        .astype("string")
        .str.strip()
        .str.lower()
        .str.replace(r"\s+", " ", regex=True)
        .replace({
            "home and kitchen": "home & kitchen"
        })
        .str.title()
    )

# Convert ratings to numeric
reviews_clean["rating"] = pd.to_numeric(
    reviews_clean["rating"], errors="coerce"
)

# Replace invalid ratings with missing values
reviews_clean.loc[
    ~reviews_clean["rating"].between(1, 5), "rating"
] = np.nan

# Fill missing ratings with the median rating
reviews_clean["rating"] = reviews_clean["rating"].fillna(
    reviews_clean["rating"].median()
)

# Standardize verified flag
if "verified_flag" in reviews_clean.columns:
    reviews_clean["verified_flag"] = (
        reviews_clean["verified_flag"]
        .astype("string").str.strip().str.title()
        .fillna("Unknown")
    )

# Convert delivery days to numeric and fill missing values with median
orders_clean["delivery_days"] = pd.to_numeric(
    orders_clean["delivery_days"], errors="coerce"
)

orders_clean["delivery_days"] = orders_clean[
    "delivery_days"
].fillna(orders_clean["delivery_days"].median())

# Standardize return flag
if "return_flag" in orders_clean.columns:
    orders_clean["return_flag"] = (
        orders_clean["return_flag"]
        .astype("string").str.strip().str.title()
        .fillna("Unknown")
    )

# Fill missing seller regions
if "region" in sellers_clean.columns:
    sellers_clean["region"] = (
        sellers_clean["region"].fillna("Unknown")
    )

# Remove rows missing essential identifiers
reviews_clean = reviews_clean.dropna(
    subset=["order_id", "product_id"]
)
products_clean = products_clean.dropna(
    subset=["product_id"]
)
orders_clean = orders_clean.dropna(
    subset=["order_id", "product_id"]
)

# Detect extreme review lengths using the IQR rule
if "review_text" in reviews_clean.columns:
    reviews_clean["review_length"] = (
        reviews_clean["review_text"].fillna("").str.len()
    )

    q1 = reviews_clean["review_length"].quantile(0.25)
    q3 = reviews_clean["review_length"].quantile(0.75)
    upper_limit = q3 + 1.5 * (q3 - q1)

    reviews_clean["extreme_review_length"] = (
        reviews_clean["review_length"] > upper_limit
    )

print("Data cleaning completed.")
print("Clean reviews:", reviews_clean.shape)
print("Clean products:", products_clean.shape)
print("Clean sellers:", sellers_clean.shape)
print("Clean orders:", orders_clean.shape)
print("Clean support:", support_clean.shape)
```

```python

valid_product_ids = set(products_clean["product_id"].dropna())

unmatched_reviews = reviews_clean[
    ~reviews_clean["product_id"].isin(valid_product_ids)
]

unmatched_orders = orders_clean[
    ~orders_clean["product_id"].isin(valid_product_ids)
]

print("Unmatched product IDs in reviews:",
      len(unmatched_reviews))
print("Unmatched product IDs in orders:",
      len(unmatched_orders))

display(unmatched_reviews.head())
display(unmatched_orders.head())
```

```python
# Merge reviews with orders
combined = pd.merge(
    reviews_clean,
    orders_clean,
    on="order_id",
    how="left",
    suffixes=("_review", "_order")
)

# Use the product ID from the review when available;
# otherwise use the order's product ID.
combined["product_id_final"] = (
    combined["product_id_review"]
    .fillna(combined["product_id_order"])
)

# Merge product details
combined = combined.merge(
    products_clean,
    left_on="product_id_final",
    right_on="product_id",
    how="left",
    suffixes=("", "_product")
)

# Merge seller details
if "seller_id" in combined.columns:
    combined = combined.merge(
        sellers_clean,
        on="seller_id",
        how="left"
    )

# Define the aggregation dictionary dynamically based on available columns
agg_dict = {}
if "complaint_category" in support_clean.columns:
    agg_dict["complaint_category"] = (
        "complaint_category",
        lambda x: ", ".join(sorted(x.dropna().astype(str).unique()))
    )
if "resolution" in support_clean.columns:
    agg_dict["resolution"] = (
        "resolution",
        lambda x: ", ".join(sorted(x.dropna().astype(str).unique()))
    )
if "ticket_id" in support_clean.columns:
    agg_dict["support_ticket_count"] = ("ticket_id", "nunique")

# Summarize support tickets by order to avoid duplicating review rows
if agg_dict:
    support_summary = support_clean.groupby("order_id").agg(**agg_dict).reset_index()
    combined = combined.merge(
        support_summary,
        on="order_id",
        how="left"
    )

print("Combined dataset shape:", combined.shape)
display(combined.head())
```

```python
print("Rating statistics:")
display(reviews_clean["rating"].describe())

print("\nRating counts:")
display(reviews_clean["rating"].value_counts().sort_index())

print("\nAverage rating by category:")
category_rating = combined.groupby(
    "category"
)["rating"].mean().sort_values()

display(category_rating)

print("\nAverage delivery days:")
print(orders_clean["delivery_days"].mean())

print("\nReturn counts:")
display(orders_clean["return_flag"].value_counts())

print("\nReviews by verification status:")
if "verified_purchase" in reviews_clean.columns:
    display(reviews_clean["verified_purchase"].value_counts())
else:
    display(reviews_clean["verified_flag"].value_counts())
```

```python

# Check column names
print("Available columns:", combined.columns.tolist())

# Feature 1: Low rating indicator
combined["low_rating"] = (
    combined["rating"] <= 2
).astype(int)

# Feature 2: Delivery category
if "delivery_days" in combined.columns:
    combined["delivery_category"] = pd.cut(
        combined["delivery_days"],
        bins=[-1, 3, 7, np.inf],
        labels=["Fast", "Normal", "Slow"]
    )
else:
    print("Warning: delivery_days column not found.")

# Feature 3: Discount category
# Find the discount column, including renamed versions
discount_col = next(
    (col for col in combined.columns
     if "discount_percent" in col.lower()),
    None
)

if discount_col:
    combined["discount_category"] = pd.cut(
        pd.to_numeric(combined[discount_col], errors="coerce"),
        bins=[-0.1, 10, 25, np.inf],
        labels=["Low", "Medium", "High"]
    )
else:
    print("Warning: Discount column not found. "
          "Check the available columns above.")

# Feature 4: Review length category
if "review_length" not in combined.columns:
    if "review_text" in combined.columns:
        combined["review_length"] = (
            combined["review_text"].fillna("").str.len()
        )
    else:
        combined["review_length"] = np.nan

combined["review_length_category"] = pd.cut(
    combined["review_length"],
    bins=[-1, 50, 150, np.inf],
    labels=["Short", "Medium", "Long"]
)

print("\nFeature engineering completed!")
display(combined.head())
```

```python

categorical_columns = combined.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

encoded_data = pd.get_dummies(
    combined,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

print("Encoded dataset shape:", encoded_data.shape)
display(encoded_data.head())
```

```python

numeric_data = combined.select_dtypes(
    include=np.number
).copy()

# Remove ID-like columns and target leakage columns
exclude_columns = [
    "rating", "low_rating"
]

feature_columns = [
    col for col in numeric_data.columns
    if col not in exclude_columns
    and not col.lower().endswith("_id")
]

correlations = numeric_data[feature_columns].corrwith(
    combined["low_rating"]
).dropna()

correlations = correlations.reindex(
    correlations.abs().sort_values(
        ascending=False
    ).index
)

print("Features most associated with low ratings:")
display(correlations.head(10))
```

```python

# Check the actual support dataset columns
print("Support dataset columns:")
print(support_clean.columns.tolist())

# Identify available columns
support_columns = support_clean.columns.tolist()

# Find the correct column names, ignoring uppercase/lowercase differences
column_map = {col.lower().strip(): col for col in support_columns}

order_col = column_map.get("order_id")
complaint_col = column_map.get("complaint_category")
resolution_col = column_map.get("resolution")
ticket_col = column_map.get("ticket_id")

# Make sure order_id exists
if order_col is None:
    raise ValueError(
        "The support dataset does not contain order_id. "
        "Check the printed column names."
    )

# Create aggregation rules only for columns that exist
agg_rules = {}

if complaint_col:
    agg_rules["complaint_category"] = (
        complaint_col,
        lambda x: ", ".join(x.dropna().astype(str).unique())
    )

if resolution_col:
    agg_rules["resolution"] = (
        resolution_col,
        lambda x: ", ".join(x.dropna().astype(str).unique())
    )

if ticket_col:
    agg_rules["support_ticket_count"] = (
        ticket_col,
        "nunique"
    )
else:
    # Count support records when ticket_id is unavailable
    agg_rules["support_ticket_count"] = (
        order_col,
        "size"
    )

# Summarize support tickets by order
support_summary = (
    support_clean
    .groupby(order_col)
    .agg(**agg_rules)
    .reset_index()
)

# Rename order column to match the main dataset
if order_col != "order_id":
    support_summary = support_summary.rename(
        columns={order_col: "order_id"}
    )

# Remove old support columns if rerunning this cell
old_support_columns = [
    col for col in [
        "complaint_category",
        "resolution",
        "support_ticket_count"
    ]
    if col in combined.columns
]

if old_support_columns:
    combined = combined.drop(columns=old_support_columns)

# Add support information to the combined dataset
combined = combined.merge(
    support_summary,
    on="order_id",
    how="left"
)

# Save and download the combined dataset
combined.to_csv("all_5_datasets_combined.csv", index=False)

print("Datasets combined successfully!")
print("Rows and columns:", combined.shape)
print("Final columns:", combined.columns.tolist())

from google.colab import files
files.download("all_5_datasets_combined.csv")
```
